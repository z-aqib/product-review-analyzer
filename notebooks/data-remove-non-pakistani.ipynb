```python
import pandas as pd
import os

# ------------- CONFIG -------------
# ALWAYS start relative to the script file
# -------------------------
# --- Detect project root (one level above "notebooks") ---
NOTEBOOK_DIR = os.getcwd()
PROJECT_ROOT = os.path.dirname(NOTEBOOK_DIR)

print("Detected project root:", PROJECT_ROOT)

# --- Define paths relative to root ---
INPUT_FILE = os.path.join(PROJECT_ROOT, "data", "raw", "amazon.csv")
OUTPUT_FILE = os.path.join(PROJECT_ROOT, "data", "raw", "pakistani_amazon.csv")

print("Reading from:", INPUT_FILE)
print("Saving to:", OUTPUT_FILE)
BLOCKED_BRANDS = [
    # all lowercase keywords to remove, e.g.:
    # "wayona", "ambrane", "portronics", ...
    "10k", "7seven", "activa", "agaro", "aircase", "airtel", "akiara", "ambrane", "amkette", "amozo", "astigo", "bajaj", "boat", "borosil", "boult", "bulfyss", "butterfly", "candes", "cello", "classmate", "cotbolt", "croma", "crompton", "crossvolt", "csi", "dealfreez", "digitek", "e cosmos", "e-cosmos", "ecosmos", "electvision", "elv", "envie", "eureka", "flix", "gizga", "goldmedal", "havells", "hb", "healthsense", "hul", "ibell", "inalsa", "instacuppa", "irusu", "isoelite", "karbonn", "kent", "king", "kitchen", "kuber", "lapster", "lava", "lifelong", "lint", "lohaya", "lunagariya", "maharaja", "morphy", "noise", "opentech", "orpat", "parker", "pigeon", "pinnaclz", "popio", "portronics", "prestige", "prettykrafts", "prushti", "ptron", "quantum", "redtech", "robustrion", "rpm", "shopoflux", "skadioo", "sounce", "storite", "striff", "sujata", "supcares", "swapkart", "synqe", "tata", "tizum", "tuarso", "tukzer", "tvara", "tygot", "urbn", "usha", "v-guard", "vu", "vw", "wayona", "wecool", "wipro", "zebronics", "zinq", "zodo", "zoul"
]
# ----------------------------------

# Load full dataset
df = pd.read_csv(INPUT_FILE)
```

```python
# # Build one big lowercase text string per row (all string columns)
# str_cols = df.select_dtypes(include=["object"]).columns
# row_text = (
#     df[str_cols]
#     .fillna("")
#     .agg(" ".join, axis=1)
#     .str.lower()
# )

# Convert only the product_name column to lowercase text for matching
row_text = df["product_name"].fillna("").str.lower()

# Start with all rows kept
keep_mask = pd.Series(True, index=df.index)

# For each blocked brand keyword, remove rows containing it (case-insensitive)
for kw in BLOCKED_BRANDS:
    kw = kw.strip().lower()
    if not kw:
        continue
    # rows that DO NOT contain this keyword
    keep_mask &= ~row_text.str.contains(kw, regex=False)

# Filter
df_clean = df[keep_mask].copy()

# Save
df_clean.to_csv(OUTPUT_FILE, index=False)

print(f"Original: {len(df)} rows")
print(f"After removing blocked brands: {len(df_clean)} rows")
print(f"Saved cleaned file to {OUTPUT_FILE}")
```
