```python
import os
import pandas as pd
import re

# -------------------------
# ALWAYS start relative to the script file
# -------------------------
# --- Detect project root (one level above "notebooks") ---
NOTEBOOK_DIR = os.getcwd()
PROJECT_ROOT = os.path.dirname(NOTEBOOK_DIR)

print("Detected project root:", PROJECT_ROOT)

# --- Define paths relative to root ---
INPUT_FILE = os.path.join(PROJECT_ROOT, "data", "raw", "pakistani_amazon.csv")
OUTPUT_FILE = os.path.join(PROJECT_ROOT, "data", "raw", "product_names.csv")

print("Reading from:", INPUT_FILE)
print("Saving to:", OUTPUT_FILE)
# ----------------------------------

# Load only the product_name column (and product_id just for reference)
use_cols = ["product_name"]
df = pd.read_csv(INPUT_FILE, usecols=use_cols)

# Optional: derive a simple "brand" as the first word in product_name
def get_brand(name: str) -> str | None:
    if pd.isna(name):
        return None
    # take part before first '(' or ','
    chunk = re.split(r'[,(]', str(name), 1)[0]
    first_word = chunk.strip().split()[0]
    return first_word

df["brand_guess"] = df["product_name"].apply(get_brand)

# Drop exact duplicates to keep it small
df_unique = df.drop_duplicates()

# Save
df_unique.to_csv(OUTPUT_FILE, index=False)

print(f"Saved {len(df_unique)} rows to {OUTPUT_FILE}")
```
