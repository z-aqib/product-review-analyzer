```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

print("1")
# 1. Model ID on Hugging Face Hub
model_id = "MuhammadHaaris/Qwen2.5-7B-ProductReviewAnalyzer-SFT-FP16"

print("2")
# 2. Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.bfloat16,  # or "auto"
    device_map="auto",
    trust_remote_code=True,  # needed if model uses custom code
)

print("3")
# 3. Prepare prompt / messages (if using chat format)
messages = [
    {"role": "system", "content": "You are an expert product review analyzer."},
    {
        "role": "user",
        "content": "Which is the best budget 43-inch 4K TV under ₹25,000 based on customer reviews?",
    },
]

print("4")
# 4. Format prompt using model’s chat-template
text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

print("5")
# 5. Tokenize and move to correct device
inputs = tokenizer([text], return_tensors="pt").to(model.device)

print("6")
# 6. Generate output
with torch.no_grad():
    generation_output = model.generate(
        **inputs, max_new_tokens=256, temperature=0.7, top_p=0.9, do_sample=True
    )

print("7")
# 7. Decode & print
response = tokenizer.batch_decode(generation_output, skip_special_tokens=True)[0]
print(response)
```
